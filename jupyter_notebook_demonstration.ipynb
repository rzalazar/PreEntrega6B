## Ejecuci車n y Streaming de Transiciones del Grafo Jer芍rquico
Visualizamos en directo la ejecuci車n del orquestador mediante el m谷todo `.stream()`.

```python
# Definici車n del estado inicial de prueba
initial_state = {
    "messages": [HumanMessage(content="Investigar la adopci車n de LangGraph en 2026 y calcular m谷tricas de impacto.")],
    "next_agent": "Supervisor",
    "research_data": None,
    "analysis_data": None,
    "step_count": 0
}

print("=" * 80)
print("EJECUTANDO ORQUESTADOR JER芍RQUICO MULTI-AGENTE")
print("=" * 80)

# Streaming del flujo de ejecuci車n nodo a nodo
for step_output in graph.stream(initial_state):
    for node_name, state_update in step_output.items():
        print(f"\n[NODO ALCANZADO]: {node_name}")
        if "next_agent" in state_update:
            print(f"   -> Transici車n decidida: {state_update['next_agent']}")
        if "research_data" in state_update:
            print(f"   Resultado Investigaci車n: {state_update['research_data'][:120]}...")
        if "analysis_data" in state_update:
            print(f"   Resultado An芍lisis: {state_update['analysis_data'][:120]}...")
        if "step_count" in state_update:
            print(f"   Contador de pasos: {state_update['step_count']}")

print("\n" + "=" * 80)
print("FLUJO FINALIZADO CORRECTAMENTE SIN BUCLES INFINITOS")
print("=" * 80)
```
